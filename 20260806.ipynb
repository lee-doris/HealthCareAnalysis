```python
import pandas as pd

# Replace with your copied Raw GitHub URL
url = "https://raw.githubusercontent.com/lee-doris/HealthCareAnalysis/refs/heads/main/enhanced_health_insurance_claims.csv"

# Load the dataset into a pandas DataFrame
df = pd.read_csv(url)

# Display the first 5 rows
df.head()
```

```python
# Overview of column types and missing values
df.info()
```

```python
# Summary statistics for numeric columns
df.describe()
```

```python
# Count of each claim status
df['ClaimStatus'].value_counts()
```

```python
# 定義您想查看的類別欄位清單
categorical_cols = [
    'PatientGender',
    'ProviderSpecialty',
    'PatientEmploymentStatus',
    'ClaimType',
    'ClaimSubmissionMethod'
]

# 迴圈印出每個欄位的類別與數量
for col in categorical_cols:
    print(f"=== {col} ===")
    print(df[col].value_counts())
    print("\n")
```

```python
# 建立樞紐表：計算 count (個數) 與 PatientAge 的 mean (平均年齡)
cross_tab = pd.pivot_table(
    df,
    index='ProviderSpecialty',
    columns='PatientGender',
    values='PatientAge',
    aggfunc=['count', 'mean']
)

# 格式化顯示：將 Count 轉成整數、Mean 四捨五入保留一位小數，並組合成「個數 (平均年齡)」格式
formatted_cross_tab = cross_tab['count'].astype(str) + " (" + cross_tab['mean'].round(1).astype(str) + ")"

# 顯示結果
formatted_cross_tab
```
